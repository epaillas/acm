# Density plot
This notebook generates plots for the density of the BGS mock catalogs.

```python
from pathlib import Path

import lsstypes
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colorbar import Colorbar

from acm.estimators.compression import Compressor, ObjectGroup

Mr = -20
target = 8.5e-3 # Mpc^-3 h^3
root = f"/pscratch/sd/s/sbouchar/acm/bgs-v2.0/mr{Mr}/measurements/base"
```

```python
pattern=r"c{cosmo_idx}_ph{phase_idx}/seed{seed}/hod{hod_idx}/density.h5"
compressor = Compressor(root=Path(root), pattern=pattern)
group = compressor.read(reader=lsstypes.read)
```

```python
def plot_density_scatter(
    group: ObjectGroup,
    param1: str,
    param2: str,
    cmap: str = "RdBu",
    **kwargs,
) -> tuple[plt.Figure, plt.Axes, Colorbar]:
    """
    Plot a scatter plot of density as a function of two parameters.

    Parameters
    ----------
    group : ObjectGroup
        The group of objects containing the density data and parameters.
    param1 : str
        The name of the first parameter to plot on the x-axis.
    param2 : str
        The name of the second parameter to plot on the y-axis.
    cmap : str, optional
        The colormap to use for the scatter plot. Default is "RdBu".
    **kwargs
        Additional keyword arguments to pass to the scatter plot.
    """
    density = np.array([obj.data for obj in group])
    p1 = np.array([obj.data.attrs[param1] for obj in group])
    p2 = np.array([obj.data.attrs[param2] for obj in group])

    fig, ax = plt.subplots(figsize=(7, 6))
    scatter = ax.scatter(p1, p2, c=density, cmap=cmap, **kwargs)
    cbar = plt.colorbar(scatter, ax=ax, label="Density (Mpc$^{-3}$ h$^3$)")
    ax.set_xlabel(param1)
    ax.set_ylabel(param2)
    return fig, ax, cbar
```

```python
fig, ax, cbar = plot_density_scatter(group, "logM_cut", "sigma", cmap="RdBu", s=5)
# TODO: Density contours
# TODO: Density interpolated lines ?
```
